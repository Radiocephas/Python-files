```python
#Basic Class Creation

class Book:

    total_books_created = 0

    def __init__(self, title, author, price):
        self.title = title
        self.author = author
        self.price = price
        Book.total_books_created += 1

    def book_info(self):
        print(f"The book \"{self.title}\" was written by {self.author} and it costs ${self.price}")

    #Add a Discount Method
    def discount_price(self, percent):
        return self.price - (self.price * percent/100)

b1 = Book("Harry Potter", "J. K. Rowlings", 29.99)
b2 = Book("New School Chemistry", "Osei Yaw Ababio", 10.99)
b3 = Book("Think Python", "Allen B. Downey", 35.50)
b4 = Book("Clean Code", "Robert C. Martin", 45.00)

books = [b1, b2, b3, b4]

for book in books:
    book.book_info()
    print(f"Discounted Price (15% off): ${book.discount_price(15):.2f}")


# Print total number of books created
print(f"Total books created: {Book.total_books_created}")
```
