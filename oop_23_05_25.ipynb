```python
class Dog:
    total_dogs = 0

    def __init__(self, name, breed):
        self.name = name
        self.breed = breed
        Dog.total_dogs += 1
    
    def bark(self):
        print (f"Woof! I'm {self.name} and I'm a {self.breed}.")
    
    @staticmethod
    def get_total_dogs():
        return f"There are {Dog.total_dogs} dogs now"
    
    def mood(self, is_hungry: bool):
        if is_hungry:
            print(f"{self.name} looks hungry. Please feed it!")
        else:
            print(f"{self.name} is happy and wagging its tail!")


dog1 = Dog("Buddy", "Golden Retriever")
dog2 = Dog("Rex", "German Shepherd")

dog1.mood(False)
dog1.mood(True)

print(Dog.get_total_dogs())  

```
