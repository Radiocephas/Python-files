```python
class Student:

    students = []

    def __init__(self, name, grade):
        self.name = name
        self.grade = grade
    
    def __str__(self):
        return f"Name: {self.name}, Grade: {self.grade}"
    
class School:
    @staticmethod
    def add_students(student):
        Student.students.append(student)

    @staticmethod
    def list_students():
        print("List of Students in this school:")
        for student in Student.students:
            print(student)

student1 = Student("Alice", "Grade 10")
student2 = Student("Brian", "Grade 11")
student3 = Student("Cynthia", "Grade 12")
student4 = Student("Daniel", "Grade 10")
student5 = Student("Ella", "Grade 11")

School.add_students(student1)
School.add_students(student2)
School.add_students(student3)
School.add_students(student4)

School.list_students()
```

```python
class Book:
    
    all_books = []

    def __init__(self, title, author, year):
        self.title = title
        self.author = author
        self.year = year
        Book.all_books.append(self)

    def __str__(self):
        return f"Title: {self.title}, Author: {self.author}, Year: {self.year}"


class Library:
    

    def __init__(self, name):
        self.name = name
        self.books = []

    def add_book(self, book):
        if book in self.books:
            print ("This book is already added to the library!")
        else:
            self.books.append(book)
            print(f"\'{book.title}\' has been added to {self.name}")

    def remove_book(self, book):
        if book in self.books:
            self.books.remove(book)
            print(f"\'{book.title}\' has been removed from {self.name}")
        else:
            print("This book is not in this library")

    def list_books(self):
        print(f"Books in {self.name}:")
        for book in self.books:
            print(book)

book1 = Book("To Kill a Mockingbird", "Harper Lee", 1960)
book2 = Book("1984", "George Orwell", 1949)
book3 = Book("Pride and Prejudice", "Jane Austen", 1813)
book4 = Book("The Great Gatsby", "F. Scott Fitzgerald", 1925)
book5 = Book("The Catcher in the Rye", "J.D. Salinger", 1951)

city_library = Library("City Library")
school_library = Library("School Library")

city_library.add_book(book1)
city_library.add_book(book2)
city_library.add_book(book2)
school_library.add_book(book3)
school_library.add_book(book4)
school_library.add_book(book5)
school_library.remove_book(book4)

city_library.list_books()
print("")
school_library.list_books()
print("")

for book in Book.all_books:
    print(book)
```

```python
class Engine:
    def __init__(self, engine_type, horsepower):
        self.engine_type = engine_type
        self.horsepower = horsepower
    
    def __str__(self):
        return f"Engine Type: {self.engine_type}, Horsepower: {self.horsepower} HP"

class Car:
    cars_created = 0
    
    def __init__(self, make, model, year, engine):
        self.make = make
        self.model = model
        self.year = year
        self.engine = engine  # Composition: Car HAS an Engine
        Car.cars_created += 1
    
    def __str__(self):
        return (f"Make: {self.make}, Model: {self.model}, Year: {self.year}, "
                f"{self.engine}")

# Create Engine instances
engine1 = Engine("V6", 300)
engine2 = Engine("Electric", 450)
engine3 = Engine("Inline-4", 180)
engine4 = Engine("V8", 520)

# Create Car instances with Engine composed
car1 = Car("Toyota", "Camry", 2018, engine1)
car2 = Car("Tesla", "Model 3", 2023, engine2)
car3 = Car("Honda", "Civic", 2020, engine3)
car4 = Car("Ford", "Mustang", 2021, engine4)

print(Car.cars_created)
print(car1)
print(car2)
print(car3)
print(car4)
```

```python
# COMPOSITION

class Battery:
    wh_capacity = 0

    def __init__(self, capacity, manufacturer) -> None:
        self.capacity = capacity
        self.manufacturer = manufacturer
        
    def __str__(self) -> str:
        return f"Capacity: {self.capacity}mAh, Manufacturer: {self.manufacturer}"
   
    def convert_pow(self):
        self.wh_capacity = (self.capacity * 3.7)/1000
        return self.wh_capacity

class Smartphone:
    def __init__(self, brand, model, battery) -> None:
        self.brand = brand
        self.model =  model
        self.battery = battery

    def __str__(self) -> str:
        return f"Brand: {self.brand}, Model: {self.model}, Battery: {self.battery}"
    
    def calc_battery_life(self, pow_cons):
        battery_life = self.battery.wh_capacity / pow_cons
        return battery_life
    
    def replace_battery(self, new_battery):
        self.battery = new_battery

battery1 = Battery(4000, "PowerCell")
battery2 = Battery(3500, "Enermax")
battery3 = Battery(4500, "VoltTech")

phone1 = Smartphone("Apple", "iPhone 13", battery1)
phone2 = Smartphone("Samsung", "Galaxy S21", battery2)
phone3 = Smartphone("OnePlus", "9 Pro", battery3)

print(battery1.convert_pow())
print(phone1.calc_battery_life(2))
print(phone1)
phone1.replace_battery(battery2)
print(phone1)
```
