```python
class Car:

    cars_created = 0
    company = "AutoMakers Inc."
    all_cars = []

    def __init__(self, make, model, year):
        self.make = make
        self.model = model
        self.year = year
        Car.cars_created += 1
        Car.all_cars.append(self)
    

    def __str__(self):
        return f"Make: {self.make}, Model: {self.model}, Year: {self.year}, Company: {Car.company}"
    
    @staticmethod
    def list_cars():
        for car in Car.all_cars:
            print(car)
        
    @classmethod
    def update_company_name(cls, new_name):
        Car.company = new_name
    
    def validate_man_year (self):
        if self.year < 1999:
            print ("This is not a valid manufacturing year!")
        else:
            print (f"The {self.make} {self.model} was manufactured in {self.year}")

car1 = Car("Toyota", "Camry", 2018)
car2 = Car("Ford", "Mustang", 1967)
car3 = Car("Tesla", "Model 3", 2023)
car4 = Car("Volkswagen", "Beetle", 1972)

print(Car.cars_created)

Car.list_cars()
```
