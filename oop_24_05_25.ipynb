```python
class Animal:

    animals = []

    def __init__(self, name, species, age):
        self.name = name
        self.species = species
        self.age = age
        Animal.animals.append(self)
    
    def __str__(self):
        return f"Name: {self.name}, Species: {self.species}, Age: {self.age}"
    
    @staticmethod
    def describe():
        print("")
        print("List of Animals:")
        for animal in Animal.animals:
            print(animal)

class Zoo:
    name = "Wildscape Haven"
    zoo_animals = []

    @staticmethod
    def add_animal(animal):
        Zoo.zoo_animals.append(animal)

    @staticmethod
    def show_animals():
        print(f"Here is the list of animals in the {Zoo.name} zoo:")
        for animal in Zoo.zoo_animals:
            print(f"Name: {animal.name}, Species: {animal.species}, Age: {animal.age}")


animal1 = Animal("Leo", "Lion", 5)
animal2 = Animal("Ella", "Elephant", 12)
animal3 = Animal("Milo", "Meerkat", 3)
animal4 = Animal("Zara", "Zebra", 7)
animal5 = Animal("Tango", "Tiger", 4)

Zoo.add_animal(animal1)
Zoo.add_animal(animal2)
Zoo.add_animal(animal3)

Zoo.show_animals()

Animal.describe()


```
