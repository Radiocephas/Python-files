```python
#print each element in a list

fruits = ["apple", "banana", "cherry", "date"]

for x in fruits:
    print(x)
```

```python
#sum all numbers in a list

numbers = [1,2, 3, 4, 5]
total = 0
for num in numbers:
    total += num

print (total)
```

```python
#sum all numbers in a list [with negative numbers]

numbers = [10, -5, 3, -2, -7]
total = 0
for num in numbers:
    total += num

print (total)
```

```python
#sum only even numbers

numbers = [1, 4, 7, 8, 10, 13]
total = 0
for num in numbers:
    if num % 2 == 0:
        total += num

print(total)
```
