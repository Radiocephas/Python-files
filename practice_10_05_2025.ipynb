```python
#Basic List Manipulation

nums = [1, 2, 3, 4, 5]

#Remove the third element

nums.pop(2)

#add 6 to the end

nums.append(6)

#insert 0 at the beginning

nums.insert(0, 0)

#Print
print(nums)
```

```python
#Create a list of squares from 1 to 10 (inclusive) using list comprehension

squares = [pow(x, 2) for x in range(1, 11)]

print(squares)
```

```python
#From this list, extract only the even numbers

numbers = [3, 6, 1, 8, 2, 5, 7]
even_numbers = [num for num in numbers if num % 2 == 0]

print(even_numbers)
```

```python
#Nested List

matrix = [[1, 2], [3, 4], [5, 6]]
target = matrix[1][1]
print(target)
```

```python
#Sort the list in reverse alphabetical order
names = ['Zara', 'Alex', 'John', 'Bella']
names.sort(reverse=True)
print(names)
```

```python
#tuple unpacking

person = ("Cephas", 24, "Ghana")
name, age, country = person
print(name)
print(age)
print(country)
```

```python
#Grocery List Manager

groceries = ['milk', 'eggs', 'bread']
groceries.append("butter")
groceries.remove("eggs")
groceries.insert(0, "apples")

print(groceries)
```

```python
#🎬 Movie Ratings

ratings = [4, 5, 3, 4, 5, 2, 5]
print(sum(ratings)/len(ratings))

rat_abv_three = [rating for rating in ratings if rating > 3]
print(rat_abv_three)
```

```python
#🚌 Bus Passenger Count

stops = [5, 3, 0, 2, 4]

print(sum(stops))

stops = [stop for stop in stops if stop != 0]

print (stops)
```

```python
#🌍 Location Coordinates

locations = [(5.6, 8.2), (1.2, 3.4), (6.6, 9.1)]
target_coord = locations[2][1]
for key, value in locations:
    print(key, value)
print(target_coord)
```

```python
#📚 Unique Book Titles

books = ["Hamlet", "1984", "Hamlet", "Dune", "Dune", "Sapiens"]
print(set(books))
print(len(books)- len(set(books)))
```

```python
#📱 App Downloads

ios_downloads = {"Facebook", "Instagram", "Twitter"}
android_downloads = {"Instagram", "Twitter", "TikTok"}

#Find the apps available on both platforms (common apps).

print(ios_downloads.intersection(android_downloads))

#Find the apps that are exclusive to Android (i.e., in Android but not iOS).

print(android_downloads - ios_downloads)

```

```python
#👥 Event Check-In

rsvped = {"Alice", "Bob", "Cephas"}
attended = {"Bob", "Diana", "Cephas"}

#Find the people who RSVP’d but didn’t attend.
print(rsvped - attended)

#Find the people who attended without RSVPing.
print(attended - rsvped)
```
