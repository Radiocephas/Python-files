```python
#Dictionary Lookup with If-Else

prices = {
    "apple": 100,
    "banana": 50,
    "mango": 150,
    # "orange" : 200
}

#Write an if-else statement that checks if "orange" is in the dictionary.
if "orange" in prices :
    print (f"The price of the orange is {prices['orange']}")
else:
    print ("Orange not found")
```

```python
#match with Dictionary Value

order = {"item": "burger", "quantity": 2}

match order["item"]:
    case "burger":
        print("burger ordered")
    case "pizza":
        print ("pizza ordered")
    case _:
        print ("Unknown order")
```

```python
#if-elif-else with Nested Dictionary

student = {
    "name": "Alex",
    "grades": {"math": 80, "english": 65}
}

math_grade = student["grades"]["math"]

if math_grade >= 90 :
    print ("Excellent in Math")
elif math_grade >= 70:
    print("Good in Math")
else:
    print ("Needs improvement in Math")
```

```python
#Nested Dictionary and If-Else Logic
employee = {
    "name": "John",
    "role": "developer",
    "performance": {"coding": 85, "communication": 75, "teamwork": 90}
}

coding_perf = employee["performance"]["coding"]

if coding_perf >= 90:
    print ("Excellent coder")
elif coding_perf >= 80:
    print("Good coder")
else:
    print("Needs imporvement")
```

```python
#match with Guards and Dictionary
product = {"name": "Laptop", "price": 900}

# p_price = product["price"]

match product["price"]:
    case p if p > 1000:
        print ("High-end product")
    case p if p <= 1000:
        print ("Budget product")
    case _:
        print ("invalid product")
```

```python
#While Loop - D-C-A-I

#Declaration
i = 1

#Condition
while i < 7:

#Action
    print(i)

#Increment
    i += 1

#with the continue keyword, the position of the action and incrementation are switched so it becomes D-C-I-[continue keyword]-A
z = 0
while z <= 10:
    z +=1
    if z % 4 == 0:
        continue
    print(z)
```

```python
i = -5
while i < 5:
    i += 1
    if i == 3:
        continue
    print(i)
```

```python
#In For loops, you can break either before or after the print

fruits = ["apple", "banana", "cherry"]
for x in fruits:
  
  if x == "banana":
    break
  print(x)
```

```python
range_value = list(range(2, 30, 3))

for x in range_value:
  print(x)

print(range_value)
```

```python
for x in range(3):
  if x == 3: break
  print(x)
else:
  print("Finally finished!")
```
