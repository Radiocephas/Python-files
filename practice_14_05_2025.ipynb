```python
def tri_recursion(k):
  if(k > 0):
    result = k + tri_recursion(k - 1)
    print(result)
  else:
    result = 0
  return result

print("Recursion Example Results:")
tri_recursion(6)
```

```python
# Write a function order_summary(item, quantity, price).
# Then call it twice:

# First with positional arguments

# Then with keyword arguments in different order

def order_summary(item, quantity, price):
    print(f"{quantity} x {item} @ ${price} each = ${quantity * price}")

# Positional arguments
order_summary("Notebook", 3, 2.5)

# Keyword arguments (out of order)
order_summary(price=2.5, item="Notebook", quantity=3)
```

```python
def factorial(n):
    if n == 0 or n == 1:
        return 1
    return n * factorial(n - 1)

print(factorial(5))  # Expected output: 120
```

```python
#Write a recursive function repeat_sentence(sentence, times) that prints the sentence repeatedly using recursion.

def repeat_sentence(sentence, times):
   if times <= 0:
      return
   print(sentence)
   repeat_sentence(sentence, times - 1)
```

```python
def reverse_string_loop(s):
    reversed_str = ""
    for char in s:
        reversed_str = char + reversed_str  # Add each character at the beginning of the result
    return reversed_str

# Test the loop function
print(reverse_string_loop("hello"))  # Output: "olleh"
```
