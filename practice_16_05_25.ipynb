```python
def countdown(n):
    if n == 0:
        print("Blast off!")
    else:
        print(n)
        countdown(n - 1)

countdown(6)
```

```python
def factorial(x):
    if x == 0:
        return 1
    else:
        return x * factorial (x - 1)
    
print(factorial(10))
```

```python
def summation(n):
    if n == 0:
        return 0
    else:
        return n + summation(n - 1)
    
print(summation(5))
```

```python
def reverse_str_loop(word):
    if len(word) == 0 or len(word) == 1:
        return word
    else:
        return reverse_str_loop(word[1:]) + word[0]
    
print(reverse_str_loop("Gordon"))
```

```python
def fib_seq(n):
    if n == 0 or n== 1:
        return n
    else:
        return fib_seq(n-1) + fib_seq(n-2)
    
print(fib_seq(5))
```

```python
def is_palindrome(s):
    # Step 1: Clean the input string
    s = ''.join(c.lower() for c in s if c.isalnum())

    # Base case: if string is empty or has 1 character
    if len(s) <= 1:
        return True

    # Recursive case: compare first and last character
    if s[0] == s[-1]:
        return is_palindrome(s[1:-1])  # check the middle part
    else:
        return False
    
print (is_palindrome("Race car"))
```

```python
def power(base, exp):
    if exp == 0:
        return 1
    elif exp == 1:
        return base
    else:
        return base * power(base, exp - 1)
    
print(power(2, 0))
```

```python
def factorial(n):
    if n == 0 or n == 1:
        return 1
    else:
        return n * factorial(n-1)

print(factorial(3))
```

```python
def gcd (num, fac):
    if fac == 0:
        return num
    else:
        return gcd(fac, num % fac)

print (gcd(100, 15))
```
